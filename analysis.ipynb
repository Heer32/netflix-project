# Wrap it in a class (using your OOP skills)



```python
class ContentAnalyzer:
    def __init__(self, dataframe):
        self.df = dataframe

    def top_genres(self, n=10):
        genres = self.df['listed_in'].str.split(', ').explode()
        return genres.value_counts().head(n)

    def content_by_year(self):
        return self.df.groupby('year_added')['show_id'].count()

    def filter_by_country(self, country):
        return self.df[self.df['main_country'] == country]
```

```python
analyzer = ContentAnalyzer(df)
analyzer.top_genres()
```

#ML PART

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Combine relevant text fields for similarity comparison
df['combined'] = df['listed_in'].fillna('') + ' ' + df['description'].fillna('')

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['combined'])

cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

def recommend(title, df=df, cosine_sim=cosine_sim):
    idx = df[df['title'] == title].index[0]
    scores = list(enumerate(cosine_sim[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:6]
    show_indices = [i[0] for i in scores]
    return df['title'].iloc[show_indices]
```

```python
recommend('Chocolate')  
```
